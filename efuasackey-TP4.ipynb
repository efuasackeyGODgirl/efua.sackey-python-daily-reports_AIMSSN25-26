Write a Python function that takes two strings and checks if they are anagrams.

```python
# Exercise 10
```

````python
def anagram(s,t):
    a = []
    b=[]
    for i in s:
        if i in t:
            a.append(True)
    for i in t:
        if i in s:
            b.append(True)
    return a == b


anagram('spar','rasp')
````
